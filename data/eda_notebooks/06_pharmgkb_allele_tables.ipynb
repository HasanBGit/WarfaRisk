# EDA: PharmGKB Allele Definition Tables (Reference)

Source: `pharmgkb_allele_tables/{VKORC1,CYP2C9,CYP4F2}_allele_definition_table.xlsx`. These are **lookup tables** (variant to star-allele/haplotype definitions), not patient rows. They matter if we ever need to translate raw genotype calls into diplotypes/phenotypes ourselves, rather than relying on pre-coded genotype columns like the ones already in the IWPC cohorts.

```python
import pandas as pd
import numpy as np
import json
from pathlib import Path

pd.set_option("display.max_columns", 100)
DATA_DIR = Path("..").resolve()
DECISIONS_PATH = Path("dataset_decisions.json")

def record_decision(name, verdict, reason, n_rows=None, n_cols=None, saved_to=None):
    decisions = {}
    if DECISIONS_PATH.exists():
        decisions = json.loads(DECISIONS_PATH.read_text())
    decisions[name] = {
        "verdict": verdict,
        "reason": reason,
        "n_rows": n_rows,
        "n_cols": n_cols,
        "saved_to": saved_to,
    }
    DECISIONS_PATH.write_text(json.dumps(decisions, indent=2))
    print(f"Recorded decision for '{name}': {verdict}")
```

```python
vkorc1 = pd.read_excel(DATA_DIR / "usable/pharmgkb_allele_tables/VKORC1_allele_definition_table.xlsx")
cyp2c9 = pd.read_excel(DATA_DIR / "usable/pharmgkb_allele_tables/CYP2C9_allele_definition_table.xlsx")
cyp4f2 = pd.read_excel(DATA_DIR / "usable/pharmgkb_allele_tables/CYP4F2_allele_definition_table.xlsx")
print("VKORC1:", vkorc1.shape)
print("CYP2C9:", cyp2c9.shape)
print("CYP4F2:", cyp4f2.shape)
```
